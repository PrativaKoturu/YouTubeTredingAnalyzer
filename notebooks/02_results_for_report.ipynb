# 02 — Figures and tables for the report

This notebook is a thin wrapper around `analytics/benchmark.py` and the graph jobs.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from streaming_algos.bloom_filter import measure_false_positive_rate
from streaming_algos.flajolet_martin import FlajoletMartin, exact_distinct
from streaming_algos.dgim import DGIM

measured, theory = measure_false_positive_rate()
print({"bloom_measured_fpr": measured, "bloom_theory_fpr": theory})

items = [f"ch-{i % 120}" for i in range(3000)]
fm = FlajoletMartin()
for x in items:
    fm.add(x)
print({"exact": exact_distinct(items), "fm": fm.estimate()})

bits = [1 if i % 4 else 0 for i in range(200)]
dgim = DGIM(50)
for b in bits:
    dgim.add_bit(b)
print({"dgim_50": dgim.estimate(50), "exact_50": sum(bits[-50:])})
```
